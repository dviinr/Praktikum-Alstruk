```python
import array

# ==========================================================
# 1. TUPLE: Menyimpan Data Tetap Panitia Inti (Immutable)
# Format: (Jabatan, Nama PIC, No. Kontak Darurat)
# ==========================================================
pic_ketua = ("Ketua Pelaksana", "Rizki", "0812-3456-7890")
pic_acara = ("Koordinator Acara", "Devi Nur Rohmah", "0831-0886-3970")
daftar_pic = [pic_ketua, pic_acara]

print("=" * 50)
print("DATA PENANGGUNG JAWAB EVENT (TUPLE)")
print("=" * 50)
for jabatan, nama, kontak in daftar_pic:
    print(f"- {jabatan:<18}: {nama} ({kontak})")


# ==========================================================
# 2. LINKED LIST: Manajemen Rundown Acara Dinamis
# Memudahkan penyisipan mata acara di tengah jadwal
# ==========================================================
class NodeAcara:
    def __init__(self, nama_sesi, durasi_menit):
        self.nama_sesi = nama_sesi
        self.durasi_menit = durasi_menit  # Disimpan dalam menit agar bisa dihitung
        self.next = None

class RundownEvent:
    def __init__(self, jam_mulai_acara="08:00"):
        self.head = None
        # Mengonversi "08:00" menjadi total menit (8 * 60 + 0 = 480 menit)
        jam, menit = map(int, jam_mulai_acara.split(":"))
        self.waktu_awal_menit = (jam * 60) + menit

    def _format_jam(self, total_menit):
        jam = (total_menit // 60) % 24
        menit = total_menit % 60
        return f"{jam:02d}:{menit:02d}"

    def tambah_sesi(self, nama_sesi, durasi_menit):
        sesi_baru = NodeAcara(nama_sesi, durasi_menit)
        if not self.head:
            self.head = sesi_baru
            return
        sekarang = self.head
        while sekarang.next:
            sekarang = sekarang.next
        sekarang.next = sesi_baru

    def sisipkan_setelah(self, target_sesi, nama_sesi_baru, durasi_menit_baru):
        sekarang = self.head
        while sekarang:
            if sekarang.nama_sesi == target_sesi:
                sesi_baru = NodeAcara(nama_sesi_baru, durasi_menit_baru)
                sesi_baru.next = sekarang.next
                sekarang.next = sesi_baru
                return
            sekarang = sekarang.next
        print(f"[Info] Sesi '{target_sesi}' tidak ditemukan.")

    def tampilkan_rundown(self):
        sekarang = self.head
        no = 1
        waktu_berjalan = self.waktu_awal_menit

        print("\n" + "=" * 50)
        print("SUSUNAN RUNDOWN ACARA (WAKTU OTOMATIS)")
        print("=" * 50)
        while sekarang:
            jam_mulai = self._format_jam(waktu_berjalan)
            waktu_selesai = waktu_berjalan + sekarang.durasi_menit
            jam_selesai = self._format_jam(waktu_selesai)

            print(f"{no}. [{jam_mulai} - {jam_selesai}] {sekarang.nama_sesi} ({sekarang.durasi_menit} mnt)")

            # Waktu mulai sesi berikutnya otomatis mengikuti waktu selesai sesi ini
            waktu_berjalan = waktu_selesai
            sekarang = sekarang.next
            no += 1

# Eksekusi Linked List Rundown
rundown = RundownEvent(jam_mulai_acara="08:00")
rundown.tambah_sesi("Registrasi Peserta", 30)
rundown.tambah_sesi("Pembukaan & Sambutan", 30)
rundown.tambah_sesi("Materi Utama: Pengenalan Kampus", 90)
rundown.tambah_sesi("Penutupan", 30)

# Menyisipkan Ice Breaking 15 menit setelah Pembukaan & Sambutan
rundown.sisipkan_setelah("Pembukaan & Sambutan", "Ice Breaking Panitia", 15)
rundown.tampilkan_rundown()


# 3. ARRAY: Rekap Kehadiran Peserta per Sesi
kehadiran_sesi = array.array('i', [145, 150, 148, 130])
total_kehadiran = sum(kehadiran_sesi)
rata_kehadiran = total_kehadiran / len(kehadiran_sesi)
sesi_terramai = max(kehadiran_sesi)

print("\n" + "=" * 50)
print("STATISTIK KEHADIRAN PER SESI (ARRAY)")
print("=" * 50)
print(f"Data Kehadiran (4 Sesi) : {kehadiran_sesi.tolist()}")
print(f"Rata-rata per Sesi      : {rata_kehadiran:.1f} peserta")
print(f"Kehadiran Tertinggi     : {sesi_terramai} peserta")
print("=" * 50)
```
