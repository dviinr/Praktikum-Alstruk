```python
# 1. Membuat variabel harga dan jumlah
harga = 12500
jumlah = 4

# 2. Menghitung total harga
total_harga = harga * jumlah

# 3. Sisa pembagian total harga dengan 2.000 (%)
sisa_pembagian = total_harga % 2000

# 4. Harga setelah diskon Rp. 5.000
harga_diskon = total_harga - 5000

# 5. Menampilkan hasil
print("Total Harga          : Rp", total_harga)
print("Sisa Pembagian (2000): Rp", sisa_pembagian)
print("Harga Setelah Diskon : Rp", harga_diskon)
```
