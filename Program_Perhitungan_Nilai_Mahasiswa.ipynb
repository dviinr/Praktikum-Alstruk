```python
#####Program Perhitungan Nilai Mahasiswa#####
# Input data mahasiswa
nama = input("Masukkan nama mahasiswa: ")
nim = input("Masukkan NIM mahasiswa: ")
nilai_tugas = float(input("Masukkan nilai Tugas: "))
nilai_uts = float(input("Masukkan nilai UTS: "))
nilai_uas = float(input("Masukkan nilai UAS: "))

# Perhitungan nilai akhir berdasarkan bobot
# Tugas: 30%, UTS: 30%, UAS: 40%
nilai_akhir = (nilai_tugas * 0.30) + (nilai_uts * 0.30) + (nilai_uas * 0.40)

# Penentuan status kelulusan
if nilai_akhir >= 60:
    status = "Lulus"
else:
    status = "Belum Lulus"

# Output hasil
print("\n" + "=" * 35)
print("HASIL NILAI MAHASISWA")
print("=" * 35)
print(f"Nama Mahasiswa  : {nama}")
print(f"NIM             : {nim}")
print(f"Nilai Akhir     : {nilai_akhir:.2f}")
print(f"Status Kelulusan: {status}")
print("=" * 35)
```
