```python
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "Kelas": "IF-3-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# 1. Tampilan nilai pertama dari list
print("Nilai pertama list :", nilai[0])

# 2. Menambahkan nilai 95 ke list
nilai.append(95)
print("List terbaru       :", nilai)

# 3. Menampilkan nama dari dictionary
print("Nama Mahasiswa     :", mahasiswa["nama"])

# 4. Tampilkan koordinat pertama dari tuple
print("Koordinat pertama  :", koordinat[0])

# 5. Tampilkan angka unik dengan sorted(angka)
print("Angka unik (sorted):", sorted(angka))
```
