```python
# 1. Membuat valiabel nilai & kehadiran
nilai = 82
kehadiran = 90

# 2. Memeriksa masing-masing syarat
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

# 3. Menggunakan operator logika and
lulus_and = syarat_nilai and syarat_kehadiran

print("Syarat Nilai (>=75)    :", syarat_nilai)
print("Syarat Kehadiran (>=80):", syarat_kehadiran)
print("Lulus (menggunakan AND):", lulus_and)

print("-"*30)

# 4. Mengubah nilai menjadi 70
nilai = 70
syarat_nilai = nilai >= 75
lulus_and = syarat_nilai and syarat_kehadiran
print("Setelah nilai diubah menjadi 70:")
print("Syarat Nilai (>=75)    :", syarat_nilai)
print("Lulus (menggunakan AND):", lulus_and)

# 5. Menggunakan operator logika or
lulus_or = syarat_nilai or syarat_kehadiran
print("Lulus (menggunakan OR):", lulus_or)
```
