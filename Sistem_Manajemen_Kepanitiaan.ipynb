```python
import array

# ==========================================================
# 1. TUPLE: Menyimpan Data Tetap Panitia Inti (Immutable)
# Format: (Jabatan, Nama PIC, No. Kontak Darurat)
# ==========================================================
pic_ketua = ("Ketua Pelaksana", "Rizki", "0812-3456-7890")
pic_acara = ("Koordinator Acara", "Devi Nur Rohmah", "0831-0886-3970")
daftar_pic = [pic_ketua, pic_acara]

print("=" * 45)
print("DATA PENANGGUNG JAWAB EVENT (TUPLE)")
print("=" * 45)
for jabatan, nama, kontak in daftar_pic:
    print(f"- {jabatan:<18}: {nama} ({kontak})")


# ==========================================================
# 2. LINKED LIST: Manajemen Rundown Acara Dinamis
# Memudahkan penyisipan mata acara di tengah jadwal
# ==========================================================
class NodeAcara:
    def __init__(self, waktu, nama_sesi):
        self.waktu = waktu
        self.nama_sesi = nama_sesi
        self.next = None

class RundownEvent:
    def __init__(self):
        self.head = None

    # Menambah sesi ke akhir rundown
    def tambah_sesi(self, waktu, nama_sesi):
        sesi_baru = NodeAcara(waktu, nama_sesi)
        if not self.head:
            self.head = sesi_baru
            return
        sekarang = self.head
        while sekarang.next:
            sekarang = sekarang.next
        sekarang.next = sesi_baru

    # Menyisipkan sesi baru setelah sesi tertentu (Dinamis)
    def sisipkan_setelah(self, target_sesi, waktu_baru, nama_sesi_baru):
        sekarang = self.head
        while sekarang:
            if sekarang.nama_sesi == target_sesi:
                sesi_baru = NodeAcara(waktu_baru, nama_sesi_baru)
                sesi_baru.next = sekarang.next
                sekarang.next = sesi_baru
                return
            sekarang = sekarang.next
        print(f"[Info] Sesi '{target_sesi}' tidak ditemukan.")

    def tampilkan_rundown(self):
        sekarang = self.head
        no = 1
        print("\n" + "=" * 45)
        print("SUSUNAN RUNDOWN ACARA (LINKED LIST)")
        print("=" * 45)
        while sekarang:
            print(f"{no}. [{sekarang.waktu}] {sekarang.nama_sesi}")
            sekarang = sekarang.next
            no += 1

# Eksekusi Linked List Rundown
rundown = RundownEvent()
rundown.tambah_sesi("08:00 - 08:30", "Registrasi Peserta")
rundown.tambah_sesi("08:30 - 09:00", "Pembukaan & Sambutan")
rundown.tambah_sesi("09:00 - 10:30", "Materi Utama: Pengenalan Kampus")
rundown.tambah_sesi("10:30 - 11:00", "Penutupan")

# Simulasi kondisi lapangan: Divisi acara menyisipkan Ice Breaking di tengah
rundown.sisipkan_setelah("Pembukaan & Sambutan", "09:00 - 09:15", "Ice Breaking Panitia")
rundown.tampilkan_rundown()


# ==========================================================
# 3. ARRAY: Rekap Kehadiran Peserta per Sesi (Homogen)
# Menyimpan jumlah peserta hadir (integer) di 4 sesi utama
# ==========================================================
kehadiran_sesi = array.array('i', [145, 150, 148, 130])

total_kehadiran = sum(kehadiran_sesi)
rata_kehadiran = total_kehadiran / len(kehadiran_sesi)
sesi_terramai = max(kehadiran_sesi)

print("\n" + "=" * 45)
print("STATISTIK KEHADIRAN PER SESI (ARRAY)")
print("=" * 45)
print(f"Data Kehadiran (4 Sesi) : {kehadiran_sesi.tolist()}")
print(f"Rata-rata per Sesi      : {rata_kehadiran:.1f} peserta")
print(f"Kehadiran Tertinggi     : {sesi_terramai} peserta")
print("=" * 45)
```
