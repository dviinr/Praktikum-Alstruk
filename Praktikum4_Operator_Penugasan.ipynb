```python
poin = 100 # Nilai awal
print("Nilai awal :", poin)

poin += 25 # Tambah 25
print("poin += 25 :", poin)

poin -= 10 # Kurangi 10
print("poin -= 10 :", poin)

poin *= 2 # Kali 2
print("poin *= 2 :", poin)

poin /= 2 # Bagi 2
print("poin /= 2 :", poin)
```
