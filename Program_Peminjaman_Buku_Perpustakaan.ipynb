```python
# Menerima input data mahasiswa dan peminjamanpeminjamanw
nama = input("Masukkan nama mahasiswa: ")
jumlah_buku = int(input("Masukkan jumlah buku yang dipinjam (maks 3): "))

# 1. Percabangan Validasi Jumlah Peminjaman Buku
if jumlah_buku > 3:
    print("\n[PERINGATAN] Jumlah peminjaman melebihi batas maksimal 3 buku!")
else:
    durasi_hari = int(input("Masukkan total hari peminjaman: "))

    # 2. Percabangan Menghitung Denda dan Status Keterlambatan
    if durasi_hari <= 3:
        status_keterlambatan = "Tepat Waktu"
        denda_per_buku = 0
        status_pinjam_lagi = "Diizinkan"
    else:
        # Menghitung hari keterlambatan (setelah hari ke-3)
        hari_terlambat = durasi_hari - 3

        # Hari pertama terlambat Rp1.000, hari berikutnya bertambah Rp2.000/hari
        denda_per_buku = 1000 + ((hari_terlambat - 1) * 2000)

        # Percabangan bersarang (Nested-IF) menentukan status sanksi
        if durasi_hari <= 7:
            status_keterlambatan = f"Terlambat {hari_terlambat} Hari (Ringan)"
            status_pinjam_lagi = "Diizinkan (Setelah Denda Lunas)"
        else:
            status_keterlambatan = f"Terlambat {hari_terlambat} Hari (Berat)"
            status_pinjam_lagi = "Dibekukan (Sanksi Keterlambatan)"

    # Total denda dihitung berdasarkan akumulasi seluruh buku yang dipinjam
    total_denda = denda_per_buku * jumlah_buku

    # 3. Menampilkan Output Nota Pengembalian
    print("\n" + "=" * 40)
    print("REKAP PEMINJAMAN BUKU PERPUSTAKAAN")
    print("=" * 40)
    print(f"Nama Mahasiswa    : {nama}")
    print(f"Jumlah Buku       : {jumlah_buku} buku")
    print(f"Total Durasi      : {durasi_hari} hari")
    print(f"Status            : {status_keterlambatan}")
    print(f"Denda per Buku    : Rp{denda_per_buku:,}")
    print(f"Total Denda       : Rp{total_denda:,}")
    print("-" * 40)
    print(f"Status Pinjam Lagi: {status_pinjam_lagi}")
    print("=" * 40)
```
